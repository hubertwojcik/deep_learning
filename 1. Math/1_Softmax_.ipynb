```python
# import libriaires
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# create data and sort it to get a smooth curve on the plot
z = torch.randint(-5, 15, (25,))
z, _ = torch.sort(z)

# Initialize softmax activation class
softfun = nn.Softmax(dim=0)

# apply the data (converting z to float)
sigmaT = softfun(z.float())

print(sigmaT)
```

```python
# show plot using z as x-axis
plt.plot(z, sigmaT, 'o-')
plt.xlabel("Original number (z)")
plt.ylabel("Probability")
plt.title("Softmax activation function")
plt.show()
```

# Enthropy

Shannona $H(p)$:

$$
H(p) = - \sum_{i=1}^{n} p_i \log_2(p_i)
$$

```python
x = [.25, .75]

H = 0

for p in x:
    H -= p * np.log(p)

print(H)


```

```python
# Cross entropy
p = [ 1,0.000001]
q = [.25, .75]

H = 0
for i in range(len(p)):
    H -= p[i] * np.log(q[i])

print(H)
```

```python
import torch
import torch.nn.functional as F

q_tensor = torch.Tensor(q)
p_tensor = torch.Tensor(p)

F.binary_cross_entropy(q_tensor, p_tensor)
```
